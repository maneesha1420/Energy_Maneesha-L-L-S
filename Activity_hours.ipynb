```python
# Program 1: Calculate total weekly energy from a list
daily_kwh = [400, 850, 400, 350, 500, 600]
total_energy = sum(daily_kwh)
print("Total Energy Used:", total_energy, "kWh")
```

```python
# Program 2: Store building usage data
building_usage = {
    "CSE": 400,
    "Hostel": 850
}
print("Hostel used:", building_usage["Hostel"], "kWh")
```

```python
# Program 3: A function to calculate electricity cost
def calculate_bill(kwh, rate_per_unit):
    return kwh * rate_per_unit

todays_bill = calculate_bill(400, 8)
print("Today's estimated bill is Rs.", todays_bill)
```

```python
# Import the pandas library
import pandas as pd

# Load the dataset you created
df = pd.read_csv('campus_energy.csv')

# Display the first few rows of the data
print("Here is the raw data:")
display(df.head())

# Group the data to see total usage per building
building_totals = df.groupby('Building')['kWh_Used'].sum()
print("\nTotal usage per building:")
print(building_totals)

# Generate a statistical summary (average, min, max) of the numeric columns
print("\nStatistical Summary:")
display(df.describe())
```

```python
# 1. Import the necessary tools (pandas for data, matplotlib for charts)
import pandas as pd
import matplotlib.pyplot as plt

# 2. Create a small practice dataset directly in the code
data = {
    'Day': ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday'],
    'CSE_Block_kWh': [450, 460, 440, 480, 430],
    'Hostel_kWh': [800, 810, 790, 820, 850]
}
df = pd.DataFrame(data)

# 3. Chart 1: A Bar Chart (CSE Block)
plt.bar(df['Day'], df['CSE_Block_kWh'], color='blue')
plt.title('CSE Block Energy Usage')
plt.ylabel('Energy (kWh)')
plt.show() # This command draws the picture on the screen

# 4. Chart 2: A Line Graph (Hostel)
plt.plot(df['Day'], df['Hostel_kWh'], color='red')
plt.title('Hostel Energy Trend')
plt.ylabel('Energy (kWh)')
plt.show()
```
